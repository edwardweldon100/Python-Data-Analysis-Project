```python
#Define Annual Income Tax function (assuming NY state rates, Federal rates for Single filers, and Federal Standard Deduction)
def annual_income_tax(salary):
    #State
    state_annual_brackets = [0, 8500, 11700, 13900, 80650, 215400, 1077550, 5000000, 25000000]
    state_bracket_pct = [.04, .045, .0525, .055, .06, .0685, .0965, .103, .109]

    state_tax = 0.0
    for i in range(len(state_annual_brackets) - 1):
        lower = state_annual_brackets[i]
        upper = state_annual_brackets[i + 1]

        if salary <= lower:
            break
        taxable_amount = min(salary, upper) - lower
        state_tax += taxable_amount * state_bracket_pct[i]

    if salary > state_annual_brackets[-1]:
        state_tax += (salary - state_annual_brackets[-1]) * state_bracket_pct[-1]

    #Federal (with standard deduction)
    federal_annual_brackets = [0, 11925, 48475, 103350, 197300, 250525, 626350]
    federal_bracket_pct = [.1, .12, .22, .24, .32, .35, .37]

    salary_w_standard_deduction = salary - 14600
    
    if salary_w_standard_deduction <= 0:
        federal_tax = 0.0
    else:
        federal_tax = 0.0
        for i in range(len(federal_annual_brackets) - 1):
            lower = federal_annual_brackets[i]
            upper = federal_annual_brackets[i + 1]

            if salary_w_standard_deduction <= lower:
                break
            taxable_amount = min(salary_w_standard_deduction, upper) - lower
            federal_tax += taxable_amount * federal_bracket_pct[i]

        if salary_w_standard_deduction > federal_annual_brackets[-1]:
            federal_tax += (salary_w_standard_deduction - federal_annual_brackets[-1]) * federal_bracket_pct[-1]

    return state_tax + federal_tax
```

```python
#Define career pay function (10 years till mid career pay)
#---assuming age 22 at start of career, 32 at mid-career, 65 at end of career.  (mid-career age can be adjusted)
#---assuming constant annual pay growth rate till end of career
#---net of income tax
def career_pay_w_4yr_degree(early_career_pay, mid_career_pay, length = '4 year', yrs_till_mid = 10):
    growth_rate = (mid_career_pay / early_career_pay) ** (1 / yrs_till_mid) - 1
    annual_pay = [early_career_pay]

    if length == '2 Year':
        for year in range(1, 45):
            year_pay = early_career_pay * (1 + growth_rate) ** year
            year_net_pay = year_pay - annual_income_tax(year_pay)
            annual_pay.append(year_net_pay)
    else:
        for year in range(1, 43):
            year_pay = early_career_pay * (1 + growth_rate) ** year
            year_net_pay = year_pay - annual_income_tax(year_pay)
            annual_pay.append(year_net_pay)
    
    return sum(annual_pay)
```

```python
#Define Present Value of career pay function (10 years till mid career pay)
#---assuming it's May of current year (graduation month)
#---assuming age 18 now, 22 at start of career, 32 at mid-career, 65 at end of career. (mid-career age can be adjusted)
#---assuming constant annual pay growth rate till end of career (applied at start of each year)
#---assuming paid bi-weekly (26 payments per year)
#---assuming 6% annual discount rate (after-tax real rate of return on S&P 500 over last 10 years)
#---net of income tax
def career_pay_w_4yr_degree_pv(early_career_pay, mid_career_pay, length = '4 year', yrs_till_mid = 10, discount_rate = 6):
    growth_rate = (mid_career_pay / early_career_pay) ** (1 / yrs_till_mid) - 1

    if length == '2 Year':
        annual_pay = [early_career_pay]
        for year in range(1, 45):
            year_pay = early_career_pay * (1 + growth_rate) ** year
            year_net_pay = year_pay - annual_income_tax(year_pay)
            annual_pay.append(year_net_pay)
        
        annual_pay = np.array(annual_pay)
        biweekly_array = np.ones(26)
        biweekly_pay = []
        for i in annual_pay:
            biweekly_pmts = biweekly_array * (i/26)
            biweekly_pay.append(biweekly_pmts)
        biweekly_pay = np.array(biweekly_pay).reshape(-1)
        
        biweekly_pay_pv_multiplier = []
        for p in range((2*26),(47*26)):
            multiplier = 1/(((1+discount_rate/100)**(1/26))**p)
            biweekly_pay_pv_multiplier.append(multiplier)
    else:
        annual_pay = [early_career_pay]
        for year in range(1, 43):
            year_pay = early_career_pay * (1 + growth_rate) ** year
            year_net_pay = year_pay - annual_income_tax(year_pay)
            annual_pay.append(year_net_pay)
        
        annual_pay = np.array(annual_pay)
        biweekly_array = np.ones(26)
        biweekly_pay = []
        for i in annual_pay:
            biweekly_pmts = biweekly_array * (i/26)
            biweekly_pay.append(biweekly_pmts)
        biweekly_pay = np.array(biweekly_pay).reshape(-1)
        
        biweekly_pay_pv_multiplier = []
        for p in range((4*26),(47*26)):
            multiplier = 1/(((1+discount_rate/100)**(1/26))**p)
            biweekly_pay_pv_multiplier.append(multiplier)
    
    career_pay_pv = np.multiply(biweekly_pay, biweekly_pay_pv_multiplier)

    career_pay_pv_total = 0
    for i in career_pay_pv:
        career_pay_pv_total = career_pay_pv_total + i
    return career_pay_pv_total
```

```python
#Define Total College Cost
def college_cost(length, annual_cost):
    if length == '2 Year':
        total_cost = 2 * annual_cost
    else:
        total_cost = 4 * annual_cost
    return total_cost
```

```python
#Define Present Value of annual college cost
#---assuming it's May of current year (graduation month)
#---assuming fall term payment due 4 months from May and and spring term payment due 4 months after fall payment
#---assuming annual tuition growth_rate of 4% and annual room and board growth rate of 4%
#---assuming 6% annual discount rate (after-tax real rate of return on S&P 500 over last 10 years)
def college_cost_pv(length, annual_cost, growth_rate = 4, discount_rate = 6):
    term_cost = annual_cost / 2
    growth_rate_num = growth_rate / 100
    if length == '2 Year':
        payments = [term_cost, term_cost, term_cost * (1 + growth_rate_num), term_cost * (1 + growth_rate_num)]
        month_till_payments_due = [4, 8, 16, 20]
        pmt_pv_multiplier = []
        for m in month_till_payments_due:
            multiplier = 1/(((1+discount_rate/100)**(1/12))**m)
            pmt_pv_multiplier.append(multiplier)
    else:
        total_cost = 4 * annual_cost
        payments_sub = [0,1,2,3]
        payment_years = [x for pair in zip(payments_sub, payments_sub) for x in pair]
        payments = []
        for i in payment_years:
            pmt = term_cost * ((1 + growth_rate_num) ** i)
            payments.append(pmt)
        month_till_payments_due = [4, 8, 16, 20, 28, 32, 40, 44]
        pmt_pv_multiplier = []
        for m in month_till_payments_due:
            multiplier = 1/(((1+discount_rate/100)**(1/12))**m)
            pmt_pv_multiplier.append(multiplier)
        
    return sum(np.multiply(payments, pmt_pv_multiplier))
```

```python
#Custom Histogram KDE Function
def custom_kde(frame_name, col_name):
    
    # Setting up the style
    sns.set_style("whitegrid")

    # Calculate Mean, Median, Mode for SalePrice
    mean_col = frame_name[col_name].mean()
    median_col = frame_name[col_name].median()
    mode_col = frame_name[col_name].mode().values[0]
     
    # Plotting the histogram
    plt.figure(figsize=(14, 7))
    sns.histplot(x=frame_name[col_name], bins=100, kde=True, color="skyblue")
    plt.axvline(mean_col, color='r', linestyle='--', label=f"Mean: {mean_col:,.0f}")
    plt.axvline(median_col, color='g', linestyle='-', label=f"Median: {median_col:,.0f}")
    plt.axvline(mode_col, color='b', linestyle='-.', label=f"Mode: {mode_col:,.0f}")
     
    # Annotations for skewness and kurtosis
    plt.annotate('Skewness: {:.2f}\nKurtosis: {:.2f}'.format(frame_name[col_name].skew(), frame_name[col_name].kurt()),
                 xy=(500000, 100), fontsize=14, bbox=dict(boxstyle="round,pad=0.3", edgecolor="black", facecolor="aliceblue"))
     
    plt.title(col_name)
    plt.legend()
    plt.xlabel(col_name)
    plt.ylabel('Frequency')
    plt.gca().set_yticklabels(['{:,.0f}'.format(y) for y in plt.gca().get_yticks()])
    plt.gca().set_xticklabels(['{:,.0f}'.format(x) for x in plt.gca().get_xticks()])

    plt.show()
```

```python
#Custom Box Plot Function
def custom_boxplot(frame_name, col_name):

    # Horizontal box plot with annotations
    plt.figure(figsize=(12, 4))
     
    # Plotting the box plot with specified color and style
    sns.boxplot(x=frame_name[col_name], color='blue', showmeans=True, meanprops={"marker": "D", "markerfacecolor": "red",
                                                                                 "markeredgecolor": "red", "markersize":10})
    
    # Titles, labels, and legends
    plt.title(col_name, fontsize=16)
    plt.xlabel(col_name, fontsize=14)
    plt.yticks([])  # Hide y-axis tick labels
    bplot.yaxis.set_major_formatter(mtick.FuncFormatter(lambda x, _: '{:,.0f}'.format(x)))
    plt.legend(handles=[Line2D([0], [0], marker='D', color='w', markerfacecolor='red', markersize=10, label='Mean')],
               loc='upper left', fontsize=14)
     
    plt.tight_layout()
    
    plt.show()
```

```python
#Custom Violin Plot Function
def custom_violinplot(frame_name, col_name):

    plot_frame = pd.DataFrame({'col': frame_name[col_name], 'frequency': np.ones(len(frame_name[col_name]))})

    repeat_values = plot_frame.loc[plot_frame.index.repeat(plot_frame['frequency'])]['col']
    
    sns.violinplot(y=repeat_values, color='#D8BFD8')
    
    plt.xlabel(col_name)
    plt.ylabel("Frequency")
    plt.title(f"Frequency Distribution for '{col_name}'")
    plt.gca().set_yticklabels(['{:,.0f}'.format(y) for y in plt.gca().get_yticks()])
    
    plt.show()
```
